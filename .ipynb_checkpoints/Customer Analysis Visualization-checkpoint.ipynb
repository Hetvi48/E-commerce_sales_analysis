```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
df = pd.read_csv('Data/4_Customer_Analysis.csv')
```

```python
df.head()
```

## Count vs Avg Order Value

```python
plt.figure(figsize=(8,3))
plt.scatter(df['Count'], df['Avg Order Value'], color='blue', alpha=0.7)
plt.title('Customer Count vs Average Order Value')
plt.xlabel('Customer Count')
plt.ylabel('Average Order Value')
plt.grid(True)
plt.show()
```

## Avg order value vs Avg rating

```python
plt.figure(figsize=(9,5))
plt.scatter(df['Avg Order Value'], df['Average Rating'], color='blue', alpha = 0.6)
plt.title('Average Order value vs Average Rating')
plt.xlabel('Avg Order Value')
plt.ylabel('Average Rating')
plt.grid(True)
plt.show()
```

## Average Delivery Time vs Average Rating

```python
plt.figure(figsize = (9,5))
plt.scatter(df['Average Delivery Time'], df['Average Rating'], color = 'blue', alpha = 0.6)
plt.title('Average Delivery Time vs Average Rating')
plt.xlabel('Average Delivery Time')
plt.ylabel('Average Rating')
plt.grid(True)
plt.show()
```
